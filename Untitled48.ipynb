```python
import time
import uuid

# ---------------------------
# Mock MCP Server (Day 2 concept)
# ---------------------------

class MockMCPServer:
    def get_tiny_image(self):
        return "[16x16 tiny image data simulated]"

mcp_server = MockMCPServer()


# ---------------------------
# Tools (Day 2 concept)
# ---------------------------

def tool_generate_image():
    """Simulates MCP tiny image generation."""
    return {"type": "image", "data": mcp_server.get_tiny_image()}

def tool_bulk_request(count):
    """Simulates a long running tool that needs approval."""
    if count <= 1:
        return {"status": "approved", "count": count}
    return {"status": "needs_approval", "count": count}


# ---------------------------
# Agent with Pause Resume (Day 2 concept)
# ---------------------------

class Agent:
    def __init__(self):
        self.state = {}
        self.memory = []

    def think(self, message):
        self.memory.append(message)

        if "generate" in message.lower():
            words = message.lower().split()
            for w in words:
                if w.isdigit():
                    count = int(w)
                    break
            else:
                count = 1

            tool_result = tool_bulk_request(count)

            if tool_result["status"] == "needs_approval":
                session_id = str(uuid.uuid4())
                self.state[session_id] = {
                    "pending": True,
                    "count": count
                }
                return {
                    "agent": "This is a large request. Please approve.",
                    "session_id": session_id,
                    "needs_confirmation": True
                }

            if tool_result["status"] == "approved":
                img = tool_generate_image()
                return {
                    "agent": f"Generated {count} image",
                    "image": img
                }

        if "list memory" in message.lower():
            return {"agent": "Here is your memory", "memory": self.memory}

        return {"agent": "I can generate images. Try: generate 3 images"}


# ---------------------------
# Pause Resume Workflow Manager (Day 2 concept)
# ---------------------------

class Workflow:
    def __init__(self, agent):
        self.agent = agent

    def send(self, text):
        response = self.agent.think(text)

        if isinstance(response, dict) and response.get("needs_confirmation"):
            print("Agent:", response["agent"])
            return response

        # Normal response
        print("Agent:", response.get("agent", ""))
        if response.get("image"):
            print("Generated image:", response["image"]["data"])
        return response

    def approve(self, session_id, confirm=True):
        session = self.agent.state.get(session_id)

        if not session:
            print("Error: No such session")
            return

        if not confirm:
            print("Agent: Request rejected")
            session["pending"] = False
            return

        count = session["count"]
        session["pending"] = False

        print("Agent: Approval received. Generating images")
        for i in range(count):
            img = tool_generate_image()
            print(f"Image {i+1}:", img["data"])


# ---------------------------
# Demo Run
# ---------------------------

agent = Agent()
flow = Workflow(agent)

# User asks to generate one image
flow.send("Generate 1 image")

# User asks to generate multiple images (needs approval)
pending = flow.send("Generate 4 images")
session_id = pending["session_id"]

# User approves
flow.approve(session_id, confirm=True)

# Check agent memory
flow.send("List memory")
```

```python
import os
from datetime import datetime, timedelta
from google.adk.agents import LlmAgent
from google.adk.runners import InMemoryRunner
from google.adk.models.google_llm import Gemini
from google.adk.tools.function_tool import FunctionTool

# Set your API key (if not already set)
# os.environ["GOOGLE_API_KEY"] = "YOUR_API_KEY_HERE"

# Mock database for storing user preferences
USER_PREFERENCES = {}

# Tool 1: Get current weather
def get_current_weather(location: str, unit: str = "celsius") -> dict:
    """
    Get current weather for a location.

    Args:
        location: City name or coordinates
        unit: Temperature unit (celsius or fahrenheit)

    Returns:
        Dictionary with weather information
    """
    mock_weather_data = {
        "new york": {"temperature": 22, "condition": "sunny", "humidity": 65},
        "london": {"temperature": 15, "condition": "cloudy", "humidity": 80},
        "tokyo": {"temperature": 18, "condition": "rainy", "humidity": 75},
        "paris": {"temperature": 19, "condition": "partly cloudy", "humidity": 70},
        "default": {"temperature": 20, "condition": "clear", "humidity": 70}
    }

    location_key = location.lower().strip()
    weather = mock_weather_data.get(location_key, mock_weather_data["default"])

    # Convert temperature if needed
    if unit.lower() == "fahrenheit":
        weather["temperature"] = (weather["temperature"] * 9/5) + 32

    return {
        "location": location,
        "temperature": weather["temperature"],
        "unit": unit,
        "condition": weather["condition"],
        "humidity": f"{weather['humidity']}%",
        "timestamp": datetime.now().isoformat()
    }

# Tool 2: Get weather forecast
def get_weather_forecast(location: str, days: int = 3) -> dict:
    """
    Get weather forecast for multiple days.

    Args:
        location: City name
        days: Number of days to forecast (1-7)

    Returns:
        Dictionary with forecast data
    """
    days = max(1, min(days, 7))  # Limit to 1-7 days

    forecast = []
    for i in range(days):
        date = datetime.now() + timedelta(days=i)
        forecast.append({
            "date": date.strftime("%Y-%m-%d"),
            "high_temp": 20 + i,
            "low_temp": 10 + i,
            "condition": ["sunny", "cloudy", "rainy"][i % 3],
            "precipitation_chance": i * 10
        })

    return {
        "location": location,
        "forecast_days": days,
        "forecast": forecast
    }

# Tool 3: Save user preferences
def save_weather_preferences(user_id: str, location: str, unit: str = "celsius") -> dict:
    """
    Save user's weather preferences.

    Args:
        user_id: Unique user identifier
        location: Preferred location
        unit: Preferred temperature unit

    Returns:
        Confirmation message
    """
    USER_PREFERENCES[user_id] = {
        "preferred_location": location,
        "preferred_unit": unit,
        "last_updated": datetime.now().isoformat()
    }

    return {
        "status": "success",
        "message": f"Preferences saved for user {user_id}",
        "preferences": USER_PREFERENCES[user_id]
    }

# Tool 4: Get user preferences
def get_user_preferences(user_id: str) -> dict:
    """
    Retrieve user's saved preferences.

    Args:
        user_id: Unique user identifier

    Returns:
        User preferences or error message
    """
    preferences = USER_PREFERENCES.get(user_id)
    if preferences:
        return {
            "status": "success",
            "preferences": preferences
        }
    else:
        return {
            "status": "error",
            "message": f"No preferences found for user {user_id}"
        }

# Tool 5: Weather comparison
def compare_weather_locations(locations: list, unit: str = "celsius") -> dict:
    """
    Compare weather between multiple locations.

    Args:
        locations: List of locations to compare
        unit: Temperature unit for comparison

    Returns:
        Comparison data
    """
    comparison = {}

    for location in locations:
        weather = get_current_weather(location, unit)
        comparison[location] = {
            "temperature": weather["temperature"],
            "condition": weather["condition"],
            "humidity": weather["humidity"]
        }

    # Find best location (simplified logic)
    best_location = max(comparison.items(),
                       key=lambda x: x[1]["temperature"] if "sunny" in x[1]["condition"] else x[1]["temperature"] - 5)

    return {
        "comparison": comparison,
        "best_location": best_location[0],
        "reason": f"Highest temperature with {best_location[1]['condition']} conditions"
    }

# Create tools
tools = [
    FunctionTool(func=get_current_weather),
    FunctionTool(func=get_weather_forecast),
    FunctionTool(func=save_weather_preferences),
    FunctionTool(func=get_user_preferences),
    FunctionTool(func=compare_weather_locations)
]

# Create agent
agent = LlmAgent(
    name="weather_agent",
    model=Gemini(model="gemini-2.0-flash-exp"),
    instruction="""You are a helpful weather assistant. Use the available tools to:
    1. Get current weather for any location
    2. Get multi-day forecasts
    3. Save and retrieve user preferences
    4. Compare weather between locations

    Be concise but helpful in your responses. When users ask about weather,
    suggest saving their preferences for future use.""",
    tools=tools
)

# Create runner
runner = InMemoryRunner(agent=agent)

print("✅ Weather Agent created successfully!")
print("Available tools:")
for tool in tools:
    print(f"  - {tool.func.__name__}")

# Async demonstration function
async def demonstrate_weather_agent():
    """Demonstrate the weather agent capabilities"""

    demo_queries = [
        "What's the weather like in New York?",
        "Get a 3-day forecast for London",
        "Save my preferences: I live in Tokyo and prefer celsius",
        "Compare weather between New York, London, and Tokyo",
        "What are my saved preferences?"
    ]

    for i, query in enumerate(demo_queries, 1):
        print(f"\n{'='*50}")
        print(f"Demo {i}: {query}")
        print(f"{'='*50}")

        try:
            # Use run_debug for simple execution
            response = await runner.run_debug(query)
            print(f"Agent: {response}")
        except Exception as e:
            print(f"Error: {e}")

# Run demonstration
print("\n🚀 Starting Weather Agent Demonstration...")
await demonstrate_weather_agent()

# Direct tool usage patterns demonstration
print(f"\n{'='*60}")
print("DIRECT TOOL USAGE PATTERNS DEMONSTRATION")
print(f"{'='*60}")

# Pattern 1: Simple data retrieval
print("\n1. Simple Data Retrieval:")
weather = get_current_weather("Paris")
print(f"Current weather in Paris: {weather}")

# Pattern 2: State management
print("\n2. State Management:")
prefs = save_weather_preferences("user123", "London", "celsius")
print(f"Saved preferences: {prefs}")

# Pattern 3: Complex operations
print("\n3. Complex Operations:")
comparison = compare_weather_locations(["New York", "London", "Tokyo"])
print(f"Weather comparison: {comparison}")

print(f"\n{'='*60}")
print("KEY CONCEPTS DEMONSTRATED:")
print("✅ External tool functions for real-time data")
print("✅ State management through user preferences")
print("✅ Complex multi-step operations")
print("✅ Error handling in tools")
print("✅ Tool interoperability")
print("✅ FunctionTool wrapper for proper tool creation")
print(f"{'='*60}")

print(f"\n{'='*60}")
print("MCP CONCEPTS IN PRACTICE:")
print("1. Tools extend agent capabilities beyond training data")
print("2. Tools can call other tools (interoperability)")
print("3. State persistence across conversations")
print("4. Consistent tool interfaces and error handling")
print("5. Real-world data integration patterns")
print(f"{'='*60}")
```
